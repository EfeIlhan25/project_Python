## Exercice 1.1 : la liste des diviseurs

Écrivez une fonction `diviseurs(n)` qui renvoie la liste des diviseurs positifs de `n`,
dans l'ordre croissant.

**Résultat attendu :** `diviseurs(28)` vaut `[1, 2, 4, 7, 14, 28]`.


```python
def diviseurs(n):
    """Liste des diviseurs positifs de n, dans l'ordre croissant."""
    resultat = []
    for d in range(1, n + 1):
        if n % d == 0:
            resultat.append(d)
    return resultat


print("TP2 - Ex 1.1 :", diviseurs(28))
```

## Exercice 2.1 : les deux PGCD

Écrivez `pgcd_naif(a, b)`, qui essaie tous les entiers de 1 à `min(a, b)`, puis
`pgcd(a, b)`, qui applique Euclide. Vérifiez qu'ils donnent le même résultat.

**Résultat attendu :** `pgcd(1071, 462)` vaut `21`, comme `pgcd_naif(1071, 462)`.


```python
def pgcd_naif(a, b):
    """PGCD en essayant tous les entiers de 1 à min(a, b)."""
    meilleur = 1
    for d in range(1, min(a, b) + 1):
        if a % d == 0 and b % d == 0:
            meilleur = d
    return meilleur


def pgcd(a, b):
    """PGCD par l'algorithme d'Euclide."""
    while b != 0:
        a, b = b, a % b
    return a


print("TP2 - Ex 2.1 :", pgcd(1071, 462), pgcd_naif(1071, 462))
print("Même résultat :", pgcd(1071, 462) == pgcd_naif(1071, 462))

def est_premier_naif(n):
    """Teste tous les diviseurs possibles de 2 à n-1."""
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True


def est_premier(n):
    """Teste seulement les diviseurs jusqu'à la racine carrée de n."""
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d = d + 1
    return True
```

## Exercice 4.1 : combien de premiers sous 100 000 ?

Comptez les nombres premiers inférieurs à 100 000, avec le crible, et mesurez le temps.
Comparez au temps qu'il faudrait en appelant `est_premier` sur chaque entier.

**Résultat attendu :** `9592` nombres premiers, obtenus en quelques centièmes de
seconde par le crible.


```python
import time

def crible(limite):
    """Renvoie la liste des nombres premiers strictement inférieurs à limite."""
    est_p = [True] * limite
    est_p[0] = False
    est_p[1] = False
    i = 2
    while i * i < limite:
        if est_p[i]:
            for multiple in range(i * i, limite, i):
                est_p[multiple] = False
        i = i + 1
    premiers = []
    for k in range(limite):
        if est_p[k]:
            premiers.append(k)
    return premiers


debut = time.perf_counter()
premiers = crible(100000)
t_crible = time.perf_counter() - debut
print("TP2 - Ex 4.1 : nombre de premiers =", len(premiers))
print("Temps du crible :", round(t_crible, 4), "s")

debut = time.perf_counter()
compteur = 0
for k in range(100000):
    if est_premier(k):
        compteur = compteur + 1
t_test = time.perf_counter() - debut
print("Avec est_premier sur chaque entier :", compteur, "en", round(t_test, 4), "s")
```

## Exercice 5.1 : factoriser

Écrivez `facteurs(n)` qui renvoie la liste des facteurs premiers de `n`, avec leurs
répétitions.

**Résultat attendu :** `facteurs(360)` vaut `[2, 2, 2, 3, 3, 5]`, et le produit de
cette liste redonne `360`.


```python
def facteurs(n):
    """Liste des facteurs premiers de n, avec répétitions."""
    resultat = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            resultat.append(d)
            n = n // d
        d = d + 1
    if n > 1:
        resultat.append(n)  
    return resultat


liste = facteurs(360)
produit = 1
for f in liste:
    produit = produit * f
print("TP2 - Ex 5.1 :", liste, "-> produit =", produit)

def fibonacci_recursif(n):
    if n < 2:
        return n
    return fibonacci_recursif(n - 1) + fibonacci_recursif(n - 2)
```

## Exercice 6.2 : comprendre pourquoi

Ajoutez à `fibonacci_recursif` un compteur d'appels, par exemple un dictionnaire qui
note combien de fois chaque valeur de `n` est demandée. Combien de fois `F(1)` est-il
recalculé pour obtenir `F(30)` ?

**Résultat attendu :** `F(1)` est calculé `832040` fois, et l'appel total déclenche
environ 2,7 millions d'appels.


```python
compteur_appels = {}


def fibonacci_compte(n):
    """Comme fibonacci_recursif, mais note chaque appel dans un dictionnaire."""
    compteur_appels[n] = compteur_appels.get(n, 0) + 1
    if n < 2:
        return n
    return fibonacci_compte(n - 1) + fibonacci_compte(n - 2)


fibonacci_compte(30)
print("TP2 - Ex 6.2 : F(1) est demandé", compteur_appels[1], "fois")
print("Nombre total d'appels :", sum(compteur_appels.values()))
```

## Exercice 7.1 : l'algorithme d'Euclide étendu

L'algorithme d'Euclide donne le PGCD ; sa version étendue donne en plus deux entiers
$u$ et $v$ tels que $au + bv = \mathrm{pgcd}(a, b)$. Écrivez
`euclide_etendu(a, b)` qui renvoie le triplet `(pgcd, u, v)`. Indication : si
`(g, u, v)` convient pour `(b, a % b)`, alors `(g, v, u - (a // b) * v)` convient pour
`(a, b)`.

**Résultat attendu :** `euclide_etendu(1071, 462)` vaut `(21, -3, 7)`, et
$1071 \times (-3) + 462 \times 7 = 21$.


```python
def euclide_etendu(a, b):
    """Renvoie (g, u, v) avec a*u + b*v = g = pgcd(a, b)."""
    if b == 0:
        return a, 1, 0
    g, u, v = euclide_etendu(b, a % b)
    return g, v, u - (a // b) * v


g, u, v = euclide_etendu(1071, 462)
print("TP2 - Ex 7.1 :", (g, u, v))
print("Vérification :", 1071 * u + 462 * v)
```

## Exercice 7.2 : l'exponentiation rapide

Calculer $a^{e} \bmod n$ en calculant d'abord $a^{e}$ est absurde quand
$e$ est grand : le nombre intermédiaire aurait des millions de chiffres. La
solution naïve fait `e` multiplications en réduisant modulo `n` à chaque tour. La
solution rapide remarque que $a^{2k} = (a^{k})^{2}$ : on élève au carré et on ne
multiplie par $a$ que lorsque le bit correspondant de $e$ vaut 1. Écrivez les
deux, `puissance_mod_naive(a, e, n)` et `puissance_mod(a, e, n)`, et comparez leurs
temps pour $7^{10^{7}} \bmod 1\,000\,003$.

**Résultat attendu :** le même résultat, `pow(a, e, n)` de Python le confirme, en une
seconde environ pour la version naïve et en une dizaine de microsecondes pour la
rapide.


```python
import time

def puissance_mod_naive(a, e, n):
    """a^e mod n avec e multiplications."""
    resultat = 1
    for i in range(e):
        resultat = (resultat * a) % n
    return resultat


def puissance_mod(a, e, n):
    """a^e mod n par exponentiation rapide (on lit les bits de e)."""
    resultat = 1
    a = a % n
    while e > 0:
        if e % 2 == 1:       
            resultat = (resultat * a) % n
        a = (a * a) % n 
        e = e // 2
    return resultat


a, e, n = 7, 10 ** 7, 1000003

debut = time.perf_counter()
r1 = puissance_mod_naive(a, e, n)
t1 = time.perf_counter() - debut

debut = time.perf_counter()
r2 = puissance_mod(a, e, n)
t2 = time.perf_counter() - debut

print("TP2 - Ex 7.2 : naïve =", r1, "en", round(t1, 3), "s")
print("rapide =", r2, "en", round(t2 * 1e6, 1), "microsecondes")
print("pow de Python =", pow(a, e, n))
```
