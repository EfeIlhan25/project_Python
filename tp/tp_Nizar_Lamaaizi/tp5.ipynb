## Exercice 1.1 : les deux conversions

Écrivez `lettre_vers_nombre(c)` et `nombre_vers_lettre(n)`, réciproques l'une de
l'autre sur les majuscules.

**Résultat attendu :** `lettre_vers_nombre("C")` vaut `2`, et
`nombre_vers_lettre(2)` vaut `"C"`.


```python
import random
from math import gcd

def lettre_vers_nombre(c):
    """'A' -> 0, 'B' -> 1, ..., 'Z' -> 25."""
    return ord(c) - ord("A")


def nombre_vers_lettre(n):
    """0 -> 'A', 1 -> 'B', ..., 25 -> 'Z'."""
    return chr(n + ord("A"))


print("TP5 - Ex 1.1 :", lettre_vers_nombre("C"), nombre_vers_lettre(2))
```

## Exercice 1.2 : nettoyer un texte

Écrivez `nettoyer(texte)` qui met tout en majuscules et ne garde que les lettres de A
à Z, en supprimant espaces, ponctuation et accents.

**Résultat attendu :** `nettoyer("Été, à Paris !")` vaut `"ETEAPARIS"`.


```python
import random
from math import gcd

ACCENTS = {"É": "E", "È": "E", "Ê": "E", "Ë": "E",
           "À": "A", "Â": "A", "Ä": "A",
           "Î": "I", "Ï": "I",
           "Ô": "O", "Ö": "O",
           "Ù": "U", "Û": "U", "Ü": "U",
           "Ç": "C"}


def nettoyer(texte):
    """Majuscules, sans accents, uniquement les lettres de A à Z."""
    resultat = ""
    for c in texte.upper():
        if c in ACCENTS:
            c = ACCENTS[c]
        if "A" <= c <= "Z":
            resultat = resultat + c
    return resultat


print("TP5 - Ex 1.2 :", nettoyer("Été, à Paris !"))
```

## Exercice 2.1 : la fonction de chiffrement

Écrivez `chiffrer(texte, a, b)` qui applique $E(x) = (ax + b) \bmod 26$ à chaque
lettre d'un texte déjà nettoyé.

**Résultat attendu :** `chiffrer("ATTAQUEALAUBE", 5, 8)` vaut `"IZZIKECILIENC"`.


```python
import random
from math import gcd

def chiffrer(texte, a, b):
    """Applique E(x) = (a*x + b) mod 26 à chaque lettre d'un texte nettoyé."""
    resultat = ""
    for c in texte:
        x = lettre_vers_nombre(c)
        resultat = resultat + nombre_vers_lettre((a * x + b) % 26)
    return resultat


print("TP5 - Ex 2.1 :", chiffrer("ATTAQUEALAUBE", 5, 8))
```

## Exercice 3.2 : l'inverse modulaire

Écrivez `inverse_modulaire(a, n)` qui renvoie l'inverse de `a` modulo `n`, ou lève une
erreur s'il n'existe pas.

**Résultat attendu :** l'inverse de 5 modulo 26 est 21, car $5 \times 21 = 105 =
4 \times 26 + 1$.


```python
import random
from math import gcd

def inverse_modulaire(a, n):
    """Inverse de a modulo n, ou erreur s'il n'existe pas."""
    for x in range(1, n):
        if (a * x) % n == 1:
            return x
    raise ValueError("pas d'inverse pour " + str(a) + " modulo " + str(n))


print("TP5 - Ex 3.2 : inverse de 5 modulo 26 =", inverse_modulaire(5, 26))
try:
    inverse_modulaire(13, 26)
except ValueError as erreur:
    print("Erreur attendue :", erreur)
```

## Exercice 4.1 : la fonction de déchiffrement

Écrivez `dechiffrer(texte, a, b)`, en refusant les clés invalides. Vérifiez que
déchiffrer ce qu'on a chiffré redonne le message d'origine.

**Résultat attendu :** `dechiffrer(chiffrer(m, 5, 8), 5, 8)` redonne `m` nettoyé.


```python
import random
from math import gcd

def dechiffrer(texte, a, b):
    """Inverse de chiffrer. Refuse les clés invalides."""
    if gcd(a, 26) != 1:
        raise ValueError("clé invalide : a doit être premier avec 26")
    inverse = inverse_modulaire(a, 26)
    resultat = ""
    for c in texte:
        y = lettre_vers_nombre(c)
        resultat = resultat + nombre_vers_lettre((inverse * (y - b)) % 26)
    return resultat


m = nettoyer("Attaque à l'aube")
secret = chiffrer(m, 5, 8)
print("TP5 - Ex 4.1 :", secret, "->", dechiffrer(secret, 5, 8))
print("Retour au message d'origine ->", dechiffrer(secret, 5, 8) == m)

texte_brut = (
    "Les mathématiques sont un langage universel que l'on apprend lentement, "
    "avec patience et méthode. Chaque problème difficile se découpe en petites "
    "étapes que l'on peut résoudre une à une. Quand on écrit un programme, on "
    "doit d'abord comprendre ce que l'on cherche, puis imaginer une méthode, et "
    "enfin vérifier que le résultat obtenu est bien celui que l'on attendait. "
    "Les nombres premiers, les polynômes et les matrices sont des outils très "
    "anciens, mais ils servent encore aujourd'hui dans presque toutes les "
    "machines que nous utilisons. Pour protéger un message, on peut remplacer "
    "chaque lettre par une autre, mais cette méthode simple est vite cassée par "
    "quelqu'un qui connaît la fréquence des lettres dans la langue française. "
    "Il existe des méthodes bien plus solides, qui mélangent plusieurs alphabets "
    "de substitution et qui demandent davantage de travail pour être attaquées. "
    "Le plus important est de ne jamais croire un résultat sans l'avoir testé, "
    "et de toujours garder un esprit curieux devant les erreurs de calcul."
)
texte_long = nettoyer(texte_brut)
secret_long = chiffrer(texte_long, 5, 8)
print("Texte long :", len(texte_long), "lettres")

message = nettoyer("Les mathématiques sont un langage universel que l'on apprend "
                   "lentement, avec patience et méthode. Chaque problème difficile "
                   "se découpe en petites étapes que l'on peut résoudre une à une.")
secret = chiffrer(message, 5, 8)
```

## Exercice 5.1 : essayer toutes les clés

Écrivez `toutes_les_clefs(secret)` qui renvoie la liste des 312 déchiffrements
possibles avec leur clé. Retrouvez le message clair à l'œil.

**Résultat attendu :** parmi les 312 propositions, une seule est du français lisible.


```python
import random
from math import gcd

def toutes_les_clefs(secret):
    """Liste de tous les déchiffrements possibles : (a, b, texte)."""
    resultat = []
    for a in range(1, 26):
        if gcd(a, 26) == 1:
            for b in range(26):
                resultat.append((a, b, dechiffrer(secret, a, b)))
    return resultat


propositions = toutes_les_clefs(secret)
print("TP5 - Ex 5.1 :", len(propositions), "propositions")
for a, b, texte in propositions[:6]: 
    print("a =", a, "b =", b, ":", texte[:40])
```

## Exercice 6.1 : casser automatiquement

Comptez les occurrences de chaque lettre du message chiffré, puis testez les clés qui
envoient E sur la lettre la plus fréquente. Retrouvez le message sans intervention
humaine.

**Résultat attendu :** le message clair, trouvé parmi une poignée de candidats au
lieu de 312.


```python
import random
from math import gcd

def compter_lettres(texte):
    """Dictionnaire {lettre: nombre d'occurrences}."""
    compteur = {}
    for c in texte:
        compteur[c] = compteur.get(c, 0) + 1
    return compteur


def casser(secret):
    """Attaque par fréquences : on suppose que E est la lettre la plus
    fréquente du clair, donc que la lettre la plus fréquente du chiffré
    est l'image de E."""
    compteur = compter_lettres(secret)
    plus_frequente = max(compteur, key=compteur.get)
    y = lettre_vers_nombre(plus_frequente)
    e = lettre_vers_nombre("E")

    candidats = []
    for a in range(1, 26):
        if gcd(a, 26) == 1:
            b = (y - a * e) % 26 
            candidats.append((a, b, dechiffrer(secret, a, b)))
    return candidats


def score(texte):
    """Compte quelques petits mots/syllabes très fréquents en français."""
    total = 0
    for motif in ("LE", "DE", "ES", "EN", "ET", "QUE", "LES", "UN"):
        total = total + texte.count(motif)
    return total


candidats = casser(secret)
print("TP5 - Ex 6.1 :", len(candidats), "candidats au lieu de 312")
meilleur = max(candidats, key=lambda c: score(c[2]))
print("Clé trouvée : a =", meilleur[0], ", b =", meilleur[1])
print("Message :", meilleur[2][:60], "...")
```

## Exercice 7.1 : l'indice de coïncidence

Écrivez `indice_coincidence(texte)`. Calculez-le sur `texte_long`, sur `secret_long`,
et sur une suite de lettres tirées au hasard de même longueur.

**Résultat attendu :** la même valeur pour le clair et pour le chiffré affine, environ
0,087 sur ce texte court, contre environ 0,039 pour les lettres au hasard.


```python
import random
from math import gcd

def indice_coincidence(texte):
    """Probabilité que deux lettres tirées au hasard dans le texte soient égales."""
    compteur = compter_lettres(texte)
    n = len(texte)
    total = 0
    for nombre in compteur.values():
        total = total + nombre * (nombre - 1)
    return total / (n * (n - 1))


random.seed(1)
hasard = ""
for i in range(len(texte_long)):
    hasard = hasard + nombre_vers_lettre(random.randint(0, 25))

print("TP5 - Ex 7.1 :")
print("texte_long  :", round(indice_coincidence(texte_long), 4))
print("secret_long :", round(indice_coincidence(secret_long), 4))
print("au hasard   :", round(indice_coincidence(hasard), 4))
```

## Exercice 8.1 : chiffrer par Vigenère

Écrivez `vigenere(texte, cle)`. Chiffrez `texte_long` avec la clé `"CLE"`, et calculez
l'indice de coïncidence du résultat.

**Résultat attendu :** un indice d'environ 0,055, intermédiaire entre le français et
le hasard. L'attaque par fréquences de l'étape 6 ne s'applique plus telle quelle.


```python
import random
from math import gcd

def vigenere(texte, cle):
    """Chaque lettre est décalée par la lettre de la clé correspondante."""
    resultat = ""
    for i in range(len(texte)):
        decalage = lettre_vers_nombre(cle[i % len(cle)])
        x = lettre_vers_nombre(texte[i])
        resultat = resultat + nombre_vers_lettre((x + decalage) % 26)
    return resultat


chiffre_vig = vigenere(texte_long, "CLE")
print("TP5 - Ex 8.1 : indice de coïncidence =",
      round(indice_coincidence(chiffre_vig), 4))

```
